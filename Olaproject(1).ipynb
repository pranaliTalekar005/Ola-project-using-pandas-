```python

import matplotlib.pyplot as plt

# First chart: Average Fare by City
avg_fare_by_grouping = df.groupby('City')['Fare'].mean()

# Second chart: Total Revenue by City
avg_revenue_by_city = df.groupby('City')['Fare'].sum().sort_values(ascending=False)

colors = ['purple', 'green', 'black', 'gray', 'pink', 'skyblue']
colors = (colors * (len(avg_revenue_by_city) // len(colors) + 1))[:len(avg_revenue_by_city)]

# Third chart: Ride Status Distribution
ride_status_by_avg = df['Ride_Status'].value_counts(normalize=True) * 100

# Create Dashboard
fig, ax = plt.subplots(2, 2, figsize=(14, 10))

# Chart 1: Average Fare
avg_fare_by_grouping.plot(kind='bar', ax=ax[0, 0], color='skyblue')
ax[0, 0].bar_label(ax[0, 0].containers[0], fmt='%.2f')
ax[0, 0].set_title('Average Fare by City')
ax[0, 0].set_xlabel('City')
ax[0, 0].set_ylabel('Average Fare')
ax[0, 0].tick_params(axis='x', rotation=45)

# Chart 2: Total Revenue
avg_revenue_by_city.plot(kind='bar', ax=ax[0, 1], color=colors)
ax[0, 1].bar_label(ax[0, 1].containers[0], fmt='%.2f')
ax[0, 1].set_title('Total Revenue by City')
ax[0, 1].set_xlabel('City')
ax[0, 1].set_ylabel('Total Revenue')
ax[0, 1].tick_params(axis='x', rotation=45)

# Chart 3: Ride Status Pie Chart
ride_status_by_avg.plot(
    kind='pie',
    ax=ax[1, 0],
    autopct='%1.1f%%',
    colors=['purple', 'skyblue', 'orange', 'pink', 'green'],
    wedgeprops={'edgecolor': 'black', 'linewidth': 2}
)
ax[1, 0].set_title('Ride Status Distribution')
ax[1, 0].set_ylabel('')

# Remove empty fourth chart
fig.delaxes(ax[1, 1])

# Dashboard Title
fig.suptitle('City Fare and Revenue Dashboard', fontsize=16)

plt.tight_layout()
plt.show()
```

```python


import pandas as pd
import plotly.express as px
from dash import Dash, dcc, html, Input, Output

# Use your existing df DataFrame

app = Dash(__name__)

cities = sorted(df['City'].dropna().unique())

app.layout = html.Div([
    html.H1(
        "Ride Analytics Dashboard",
        style={'textAlign': 'center'}
    ),

    html.Label("Select City"),

    dcc.Dropdown(
        id='city-dropdown',
        options=[
            {'label': 'All Cities', 'value': 'All'}
        ] + [
            {'label': city, 'value': city}
            for city in cities
        ],
        value='All',
        clearable=False
    ),

    dcc.Graph(id='average-fare-chart'),
    dcc.Graph(id='total-revenue-chart'),
    dcc.Graph(id='ride-status-chart')
])


@app.callback(
    Output('average-fare-chart', 'figure'),
    Output('total-revenue-chart', 'figure'),
    Output('ride-status-chart', 'figure'),
    Input('city-dropdown', 'value')
)
def update_dashboard(selected_city):

    # Filter data by City
    if selected_city == 'All':
        filtered_df = df.copy()
    else:
        filtered_df = df[df['City'] == selected_city]

    # Chart 1: Average Fare by City
    avg_fare = (
        filtered_df.groupby('City', as_index=False)['Fare']
        .mean()
    )

    fig1 = px.bar(
        avg_fare,
        x='City',
        y='Fare',
        title='Average Fare by City',
        color='City',
        text_auto='.2f'
    )

    # Chart 2: Total Revenue by City
    revenue = (
        filtered_df.groupby('City', as_index=False)['Fare']
        .sum()
        .sort_values('Fare', ascending=False)
    )

    fig2 = px.bar(
        revenue,
        x='City',
        y='Fare',
        title='Total Revenue by City',
        color='City',
        text_auto='.2f'
    )

    # Chart 3: Ride Status Distribution
    status = (
        filtered_df['Ride_Status']
        .value_counts()
        .rename_axis('Ride_Status')
        .reset_index(name='Count')
    )

    fig3 = px.pie(
        status,
        names='Ride_Status',
        values='Count',
        title='Ride Status Distribution',
        hole=0.3
    )

    return fig1, fig2, fig3


if __name__ == '__main__':
    app.run(debug=True)
```
